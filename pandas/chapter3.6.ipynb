```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

tips = sns.load_dataset("tips")
```

```python
scatter = sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",
    size="size",  # ⭐️ 直接将列名传给 size 参数
    sizes=(20, 200),  # 可选：设置点的大小范围
    kind="scatter",  # 明确指定是散点图
)

plt.show()
```

```python
# ⭐️ 替代方案：使用纯线形标记 '+' (加号) 和 'x' (叉)
scatter_plot = sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",
    style="sex",
    markers=["+", "x"],  # '+' 和 'x' 都是线形艺术，不会有填充冲突
    size="size",
    sizes=(20, 200),
    kind="scatter",
)

plt.show()
```

```python
anscombe = sns.load_dataset("anscombe")

sns.lmplot(
    x="x",
    y="y",
    data=anscombe,
    fit_reg=True,
    col="dataset",
    col_wrap=3,
)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

# 1. 初始化 FacetGrid，按 "time" 分列
facet = sns.FacetGrid(tips, col="time")


# 3. 映射核密度估计图 (KDE Plot)
facet.map(
    sns.kdeplot,
    "total_bill",
    fill=True,
    color="red",
    alpha=1,
)

# 4. 映射地毯图 (Rugplot)
facet.map(sns.rugplot, "total_bill")

facet.map(
    sns.histplot,
    "total_bill",
    stat="density",
    alpha=0.5,
)

plt.show()
```

```python
facet = sns.FacetGrid(tips, col="day", hue="sex")
facet = facet.map(plt.scatter, "total_bill", "tip")
facet = facet.add_legend()
plt.show()
```

```python
fig = sns.lmplot(
    x="total_bill", y="tip", data=tips, fit_reg=False, hue="sex", col="day"
)
plt.show()
```
